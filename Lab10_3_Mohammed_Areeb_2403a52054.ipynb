Visualizing word embeddings using t-SNE is a technique used to reduce high-dimensional word vectors (e.g., 300 dimensions) into 2D or 3D space for visualization. Word embeddings generated by models like Word2Vec or GloVe capture semantic relationships between words. t-SNE preserves local similarity and clusters similar words together in lower dimensions, helping in understanding semantic structure and validating embedding quality.

```python
import gensim.downloader as api   # To load pretrained word embeddings
import numpy as np                # For numerical operations
import pandas as pd               # For structured data handling (optional but useful)
import matplotlib.pyplot as plt   # For plotting
from sklearn.manifold import TSNE # For dimensionality reduction
```

```python
# Load pre-trained Word2Vec model (100D version for speed)
model = api.load("glove-wiki-gigaword-100")

# Print vocabulary size
print("Vocabulary Size:", len(model))

# Display one example vector
print("Vector for 'king':")
print(model['king'])
```

Themes Chosen for 5  to 6 lines

Animals

Cities

Technology

Royal Family

These themes were selected because they represent clearly distinct semantic groups. Animals share biological relationships, cities share geographical similarity, technology words relate to devices and computing, and royal terms are semantically connected through hierarchy. Using diverse themes helps visually confirm whether embeddings properly capture meaning and clustering behavior. By selecting categories that are conceptually different from one another, it becomes easier to observe whether t-SNE forms separate clusters in the visualization. If the embeddings are meaningful, words within the same theme should appear closer together compared to words from different themes. This structured selection also helps in clearly identifying overlaps, outliers, or unexpected placements. Overall, using diverse and well-defined themes improves the clarity and interpretability of the visualization results.

```python
words = [
    # Animals
    "dog","cat","lion","tiger","wolf","elephant","horse","monkey","zebra","bear",

    # Cities
    "paris","london","tokyo","mumbai","delhi","newyork","beijing","rome","dubai","berlin",

    # Technology
    "computer","laptop","keyboard","mouse","internet","software","hardware","python","google","facebook",

    # Royal
    "king","queen","prince","princess","monarch","royal","crown","throne","duke","empire"
]

vectors = np.array([model[word] for word in words])
print("Shape of vector matrix:", vectors.shape)

tsne = TSNE(n_components=2, random_state=42, perplexity=5)
reduced_vectors = tsne.fit_transform(vectors)

print("Shape after reduction:", reduced_vectors.shape)
```

```python
plt.figure(figsize=(12,8))

for i, word in enumerate(words):
    x, y = reduced_vectors[i]
    plt.scatter(x, y)
    plt.text(x+0.01, y+0.01, word, fontsize=9)

plt.title("t-SNE Visualization of Word Embeddings")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.show()
```

The t-SNE visualization shows clear clustering among semantically related words. Animal words such as dog, cat, and lion appear close together, indicating similar biological meaning. City names form another visible cluster, suggesting geographical similarity captured by embeddings. Technology-related terms like computer, software, and internet are grouped together. Royal terms such as king, queen, and prince also appear near each other, demonstrating hierarchical semantic relationships. Some overlap between categories may occur due to contextual usage in text data. For example, empire may appear slightly separated due to broader contextual meaning. Overall, the visualization confirms that embeddings capture semantic similarity effectively.
